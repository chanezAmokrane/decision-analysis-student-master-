# Description de la Population d'Étude

```python
def summarize_categorical(df, column_name):
    summary = pd.DataFrame(df[column_name].value_counts())
    summary.columns = ['Effectifs']
    summary['Pourcentage'] = (df[column_name].value_counts(normalize=True) * 100).round(2)
    summary = summary.sort_index().reset_index()
    summary.columns = ['Catégorie', 'Effectifs', 'Pourcentage']  # Renommage direct de 'index'
    summary['Variable'] = column_name
    summary = summary[['Variable', 'Catégorie', 'Effectifs', 'Pourcentage']]  # Réordonner les colonnes
    return summary


# Liste des variables à résumer
variables = ['Âge', 'Sexe', 'NiveauÉtudesActuel', 'DomaineEtude']

# Création du DataFrame de résumé
summary_df = pd.concat([summarize_categorical(df, var) for var in variables]).reset_index(drop=True)

# Remplacer les noms de variables répétés par une chaîne vide
last_var = None
for idx, row in summary_df.iterrows():
    if row['Variable'] == last_var:
        summary_df.at[idx, 'Variable'] = ''
    else:
        last_var = row['Variable']

# Renommer la colonne 'index' pour refléter son contenu (e.g., 'Catégorie')
summary_df.rename(columns={'index': 'Catégorie'})

summary_df.style.format({'Pourcentage': '{:.2f}%'}).background_gradient(subset=['Pourcentage'], cmap='viridis')




```

**Hiérarchie des critères et alternativess**

```python
import networkx as nx
import matplotlib.pyplot as plt

# Define your criteria and alternatives if not already done
criteria_abbreviations = {
    'Personnel': 'Personnel',
    'Réputation': 'Réputation',
    'Famille': 'Famille',
    'Métier': 'Métier',
    'Résultats': 'Résultats',
    'Matières': 'Matières',
    'Coût': 'Coût'
}

# Title of the objective node
titre_objectif = "Déterminer l'importance relative des facteurs influençant le choix de spécialisation en master chez les étudiants"

# Create the graph
G = nx.DiGraph()

# Add the objective node
G.add_node(titre_objectif)

# Add criteria nodes and edges from the objective node to criteria nodes
for crit in criteria_abbreviations.values():
    G.add_node(crit)
    G.add_edge(titre_objectif, crit)

# Add alternative nodes and edges from criteria nodes to alternative nodes
for i in range(1, 11):
    alt = f"E{i}"
    G.add_node(alt)
    for crit in criteria_abbreviations.values():
        G.add_edge(crit, alt)

# Define positions for the nodes
pos = {titre_objectif: (0, 2)}  # Objective node at the top
# Arrange criteria nodes just below the objective node
pos.update((crit, (i, 1)) for i, crit in enumerate(criteria_abbreviations.values()))
# Arrange alternatives nodes at the bottom
pos.update((f"E{i}", (i - 0.5 * (10 - 1), 0)) for i in range(1, 11))

# Draw the graph
plt.figure(figsize=(15, 8))
# Draw the objective node differently, with a 's' shape and a special color
nx.draw_networkx_nodes(G, pos, nodelist=[titre_objectif], node_shape='s', node_color='none', node_size=5000)
# Draw criteria nodes
nx.draw_networkx_nodes(G, pos, nodelist=criteria_abbreviations.values(), node_color='skyblue', node_size=3000)
# Draw alternative nodes
nx.draw_networkx_nodes(G, pos, nodelist=[f"E{i}" for i in range(1, 11)], node_color='skyblue', node_size=2000)
# Draw edges
nx.draw_networkx_edges(G, pos, arrows=True)
# Draw labels
nx.draw_networkx_labels(G, pos)
plt.title('Hiérarchie AHP des Critères et Alternatives', size=15)
plt.axis('off')  # Turn off the axis
plt.tight_layout()  # Adjust layout
plt.show()
```

**Comparaisons par paires des critères et alternatives.**

```python
import pandas as pd
import numpy as np

# Création de la matrice de comparaison par paires
comparison_matrix = np.array([
    [1.00, 3.00, 3.00, 5.00, 5.00, 7.00, 9.00],
    [0.33, 1.00, 1.00, 1.00, 3.00, 5.00, 7.00],
    [0.33, 1.00, 1.00, 1.00, 3.00, 5.00, 7.00],
    [0.20, 1.00, 1.00, 1.00, 3.00, 5.00, 7.00],
    [0.20, 0.33, 0.33, 0.33, 1.00, 3.00, 5.00],
    [0.14, 0.20, 0.20, 0.20, 0.33, 1.00, 3.00],
    [0.11, 0.14, 0.14, 0.14, 0.20, 0.33, 1.00]
])

# Création d'un DataFrame pour représenter la matrice sous forme de tableau
df = pd.DataFrame(
    comparison_matrix,
    columns=["Personnel", "Métier", "Matière", "Coûte", "Réputation", "Résultat", "Famille"],
    index=["Personnel", "Métier", "Matière", "Coûte", "Réputation", "Résultat", "Famille"]
)

# Affichage du tableau avec les bordures
print(df)
```

**vecteurs propres pour obtenir les poids des critères**

```python

# Calcul du vecteur de priorités en prenant la moyenne géométrique des lignes, puis en normalisant
priority_vector = np.prod(comparison_matrix, axis=1) ** (1 / len(comparison_matrix))
priority_vector_normalized = priority_vector / np.sum(priority_vector)
print("Vecteur de priorités normalisé:", priority_vector_normalized)

# Calcul de lambda_max, qui est une mesure de la cohérence de la matrice
lambda_max = np.mean(np.sum(comparison_matrix * priority_vector_normalized, axis=1) / priority_vector_normalized)
print("Valeur propre maximale (λ_max):", lambda_max)

# Détermination de l'indice de cohérence (CI), qui évalue à quel point les jugements sont cohérents
n = len(comparison_matrix)
CI = (lambda_max - n) / (n - 1)
print("Indice de cohérence (CI):", CI)

# Utilisation des valeurs RI pré-définies pour calculer le rapport de cohérence (CR)
RI_dict = {1: 0, 2: 0, 3: 0.58, 4: 0.90, 5: 1.12, 6: 1.24, 7: 1.32, 8: 1.41, 9: 1.45, 10: 1.49}
RI = RI_dict[n]
CR = CI / RI
print("Rapport de cohérence (CR):", CR)

# Si le CR est inférieur à 0.1, la matrice est considérée comme cohérente.
if CR < 0.1:
    print("La matrice est cohérente.")
else:
    print("La matrice n'est pas cohérente, envisagez de réviser les comparaisons.")
```
